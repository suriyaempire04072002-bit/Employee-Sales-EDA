## 1. Load the dataset into a Pandas DataFrame

```python
data=[[101,25,"F","HR",2,22000,150000],
      [102,30,"M","Sales",5,35000,320000],
      [103,28,"F","IT",4,4000,28000],
      [104,35,"M","Sales",8,55000,450000],
      [105,40,"M","HR",10,40000,200000],
      [106,29,"F","IT",3,38000,300000],
      [107,32,"M","Sales",6,48000,410000],
      [108,27,"F","HR",2,21000,160000]]

data=pd.DataFrame(data,columns=['Emp_ID','Age','Gender','Department','experience','salary','sales'])
data
```

```python
data = pd.DataFrame(data, columns=['Emp_ID','Age','Gender','Department','experience','salary','sales'])
data
```

## 5. Find the average salary and total sales per department

```python
data["salary"].mean()
print(f"Average salary = {data["salary"].mean()}")


data['salary'].mean()
print(f"Avg salary = {data.groupby("Department")['salary'].mean()}")


data.groupby('Department')['sales'].sum()
print(f"Totalsales = {data.groupby('Department')['sales'].sum()}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(6, 4))
plt.boxplot(data['salary'], patch_artist=True)
plt.title("Boxplot - Salary Outlier Detection")
plt.ylabel("Salary")
plt.show()
```

## 8.Plot a histogram for Age distribution

```python
plt.Figure(figsize=(8,5))
plt.hist(data['Age'], color='steelblue', edgecolor='black')
plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.tight_layout()
plt.show()
```

## 9 Plot a bar chart for department-wise average salary.


```python
avgsalary = data.groupby("Department")['salary'].mean()
plt.figure(figsize=(8,6))
plt.bar(avgsalary.index,avgsalary.values,color=['blue','green','yellow'],edgecolor =['black'])
plt.title('department-wise average salary')
plt.xlabel('department')
plt.ylabel('salary')
plt.tight_layout()
plt.show()

```

## 10. Plot a scatter plot between Experience and Salary

```python
plt.figure(figsize=(8,6))
plt.scatter(data['experience'],data['salary'],color=['steelblue'],edgecolors=['green'],s=102)
plt.title("experience vs salary")
plt.xlabel("experience")
plt.ylabel("salary")
plt.show()
```

## 11. Check correlation between Salary and Sales.

```python
correlation = data['salary'].corr(data['sales'])
print(f"correlation={correlation}")
```

```python
correlation=data[['salary','sales']].corr()
print("correlation",correlation)
```
